```python
import pandas as pd
titanic = pd.read_csv("https://peter.butka.website.tuke.sk/test/titanic2.csv")
titanic.head()
```

```python
titanic.info()
```

```python
titanic['age_ordinal'] = titanic['age_ordinal'].astype('category')
unique_values = titanic['fare_ordinal'].unique()
titanic['fare_ordinal'] = pd.Categorical(titanic['fare_ordinal'], categories=unique_values)
```

```python
titanic['dAge'] = pd.cut(titanic['age'], bins=[0, 18, 100], labels=['child', 'adult'])
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
titanic['fare'] = scaler.fit_transform(titanic[['fare']])
titanic['age'] = scaler.fit_transform(titanic[['age']])
```

```python
import numpy as np
ind = np.random.choice(2, size=len(titanic), replace=True, p=[0.8, 0.2])
train_data = titanic[ind == 1]
test_data = titanic[ind == 0]
```

```python
train = train_data[['fare', 'age']]
test = test_data[['fare', 'age']]
trainS = train_data['survived'].astype('category')
testS = test_data['survived'].astype('category')
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(train, trainS)
```

```python
predict = model.predict(test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(np.reshape(testS,-1), predict))
```
