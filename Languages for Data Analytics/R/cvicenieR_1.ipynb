### Vektory (Vector)
V prvej časti sa pozrieme akými spôsobmi sa dajú vektory vytvoriť. V druhej časti budeme pristupovať k jednotlivým elementom vektora podľa pozície a hodnoty, ktorú chceme vrátiť.

```r
# vytvorenie ciselnych vektorov

v = c(5.5, 6.2, 7.8)  
 v
v = 10:30              
v
v = seq(1,5, by=0.2)  
v
v = rep(1:4, times=3) 
v
v = rep(1:4, each=3)  
v
print(v) 
```

```r
# vytvorenie vektora logickych hodnot
v = c(TRUE, FALSE, FALSE) 
print(v)
```

```r
# vytvorenie vektora znakov
v = c("ab", "cd", "ef")   
print(v)
```

```r
vector = c(1, 6, 9, 12, 58, 13, 11, 89, 9)

# 1. PODLA POZICIE
print(vector[1])         # vrati prvy znak vektora

print(vector[-1])        # vrati vsetky znaky vektora okrem preveho

print(vector[1:4])       # vrati prve 4 znaky

print(vector[-(1:4)])    # vrati vsetky znaky okrem prevých styroch

print(vector[vector>6])  # vrati vacsie ako 6
```

```r
# 2. PODLA HODNOTY
print(vector[vector == 9])  # vrati hodnoty rovné 9
print(vector[vector < 12]) # vrati hodnoty menšie ako 12
```

### Zmena dátového typu (Data Type) - explicitná korekcia
Poznáme niekoľko základných dátových typov a to: znaky (`character`), logické hodnoty (`logical`), numerické hodnoty (`numeric`), celočíselné hodnoty (`integer`), komplexné čísla (`complex`) a špeciálny dátový typ faktor (`factor`). Faktor sa používa na reprezentáciu kategorických dát. Niektoré algoritmy si vyžadujú zmenu dátového typu, preto si spôsob ako "pretypovať" nápriklad vektor hodnôt.



```r
# uprava typu vektora - explicitna korekcia
z = 0:9  
z
class(z)
# zmena datoveho typu na numericky
n = as.numeric(z)
print(n)
class(n)
```

```r
# zmena datoveho typu na logický - TRUE, FALSE
l = as.logical(z)
print(l)
class(l)
```

```r
# zmena datoveho typu na vektor znakov
ch = as.character(z) 
print(ch)
class(ch)
```

```r
# zmena datoveho typu na faktory
f = as.factor(z)
print(f)
class(f)
```

### Zoznamy (List)
Zoznam je opäť špecialný objekt typu `vector`, avšak môže obsahovať rôzne dátové typy a objekty (napríklad znaky, čisla, logické hodnoty ale aj vnorený zoznam či vektor)

```r
zoznam = list(a1 = 1:3, a2 = 0.3, a3 = "d",  a4=list(c(1,3),TRUE))
print(zoznam)
```

```r
# 1. vypise prvky elementu !!!! typ integer !!!!
z1 = zoznam[[1]]  # na prvej pozicii
print(z1)
class(z1)
```

```r
z2 = zoznam$a1    # z nazvom elementu a2
print(z2)
class(z2)
```

```r
# 2. vytvori NOVY ZOZNAM  !!!! typ list !!!!, ktory bude obsahovat:
z3 = zoznam["a2"]   # prvky elementu z nazvom a2
print(z3)
class(z3)
```

```r
z4 = zoznam[2]      # prvky elementu na druhej pozicii
print(z4)
class(z4)
```

```r
z5 = zoznam[c(1,3)] # prvky prveho a tretieho elementu povodneho zoznamu
print(z5)
class(z5)
```

### Cyklus for
Cyklus s fixným počtom opkovaní:

```r
# FOR
f = seq(1,100, by = 2)
a = NULL

for (i in 1:50)
{
  a[i] = f[i]^2
}
print(a)
```

```r
# for, next 
for(i in 40:60)
{
  if(i == 50)
  {
    next
  }
  print(i)
}
```

### Cyklus while
Cyklus, ktorý sa opakuje, pokiaľ je platná podmienka:

```r
count = 0

while(count < 10)
{
  print(count)
  count = count + 1   
}
```

### Break, repeat, next, return

| Príkaz | Popis |
| --- | --- |
|`repeat` | spúšťa nekonečný cyklus, ako while, bez podmienky |
|`break`  | ukončuje beh nekonečného cyklu |
|`next`   | preskočí iteráciu v cykle |
|`return` | výstup funkcie |

```r
sum = 1 

repeat
{
  sum = sum + 2
  print(sum)
  if (sum > 11)
    break
}
```

#### Úloha 1.1

Vytvorte funkciu s názvom `funkcia`, ktorá bude mať vstupný parameter `n`. Použitím cyklu prejdite všetky čísla od `1` po `n` a pomocou vetvenia urobte nasledovne operácie:

1. Ak sa jedná o číslo deliteľné 2 jeho druhú mocninu,
2. Ak sa jedná o číslo deliteľné 3 a zároveň 5 jeho tretiu mocninu.
3. Ak číslo nepatrí ani do jednej skupiny, tak ho iba vypíšte.

Na záver spustite funkciu s názvom funkcia.

```r
funkcia = function(n){
    for(i in 1:n){
        if(i%%2==0){
            print(i^2)
        }
        else if(i%%3==0 & i%%5==0){
            print(i^3)
        }
        else{
            print(i)   
        }
    }
}
funkcia(10)
```
