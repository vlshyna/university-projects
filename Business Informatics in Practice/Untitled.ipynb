```python
mean_rate_kitchen = data['rate_kitchen'].mean()
mean_rate_laundries = data['rate_laundries'].mean()
mean_rate_common_areas = data['rate_common_areas'].mean()
mean_rate_internet = data['rate_internet'].mean()
mean_rate_police = data['rate_police'].mean()
mean_rate_events = data['rate_events'].mean()

print("Mean value for rate_kitchen:", mean_rate_kitchen)
print("Mean value for rate_laundries:", mean_rate_laundries)
print("Mean value for rate_common_areas:", mean_rate_common_areas)
print("Mean value for rate_internet:", mean_rate_internet)
print("Mean value for rate_police:", mean_rate_police)
print("Mean value for rate_events:", mean_rate_events)
```

```python
import matplotlib.pyplot as plt

variables = ['rate_kitchen', 'rate_laundries', 'rate_common_areas', 'rate_internet', 'rate_police', 'rate_events']

mean_values = [mean_rate_kitchen, mean_rate_laundries, mean_rate_common_areas, mean_rate_internet, mean_rate_police, mean_rate_events]

plt.figure(figsize=(10, 6))
plt.bar(variables, mean_values, color='skyblue')
plt.xlabel('Variables')
plt.ylabel('Average value')
plt.title('Comparison of average values')
plt.grid(axis='y', linestyle='--', alpha=0.7)  
plt.tight_layout() 
plt.show()
```
