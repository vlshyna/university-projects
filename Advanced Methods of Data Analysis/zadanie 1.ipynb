Na vstupe máte textový súbor, ktorý má na každom riadku dve slová oddelené medzerou. Vašou úlohou je:

Načítať text po riadkoch a oddeliť jednotlivé slová.
Rozdeliť slová na znaky a vytvoriť index znakov, tzn. mapu ktorá priradí každému znaku, ktorý sa vyskytoval v slovách index od 1 - počet rôznych znakov.
Vytvoriť dva zoznamy, jeden pre všetky prvé slová a jeden pre všetky druhé slová na riadku, ktoré obsahujú zoznamy indexov znakov.
Napr. ak je na vstupe text:

otcovi otec

bratovi brat

bratov brat

Ak je index {'o': 1, 't': 2, 'c': 3, 'v': 4, 'i': 5, 'e': 6, 'b': 7, 'r': 8, 'a': 9}, potom na výstupe bude:

first = [[1, 2, 3, 1, 4, 5], [7, 8, 9, 2, 1, 4, 5], [7, 8, 9, 2, 1, 4]]
second = [[1, 2, 6, 3], [7, 8, 9, 2], [7, 8, 9, 2]]

```python
#Načítať text po riadkoch a oddeliť jednotlivé slová.
with open("z1.txt") as file:
    lines = [line.split() for line in file]

print(lines)
```

```python
#Rozdeliť slová na znaky
letters = []
for line in lines:
        pair = [list(word) for word in line]
        letters.append(pair)
print(letters)
```

```python
# vytvoriť index znakov
idx_chr = set()
for line in letters:
    for word in line: 
        idx_chr.update(word)
```

```python
print(idx_chr)
```

```python
index = 1
idx = {}
for char in idx_chr:
    idx[char] = index
    index +=1
```

```python
print(idx)
```

```python
#Vytvoriť dva zoznamy
first_words = []
second_words = []
```

```python
for line in letters:
    first = [idx[ch] for ch in line[0]]
    second = [idx[ch] for ch in line[1]]
    first_words.append(first)
    second_words.append(second)
```

```python
print(first_words)
print(second_words)
```
